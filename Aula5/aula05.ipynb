```python
import pandas as pd

# Criar a DATA  FRAME

df = pd.read_csv('tabela-fipe-historico-precos.csv')

display(df.head(10))

print(df.shape)

print(df.info())
```

```python
#  Qual o carro mais caro de toda a série histórica

# Função "loc" localiza não apenas o campo, mas o registro inteiro
maisCaro = df.loc[df['valor'].idxmax()]

print(maisCaro)
```

```python
# nunique - conta registros "ÚNICOS"

# quantas marcas já operaram no Brasil.
# quantas modelos já operaram no Brasil.

qtdeMarcas = df['marca'].nunique()
print(f'Entre 1992 até 2022 {qtdeMarcas} marcas operam no Brasil')

qtdeModelo = df['modelo'].nunique()
print(f'Entre 1992 até 2022 {qtdeModelo} modelos operam no Brasil')
```

```python
# unique - apesenta apenas registros únicos

marcas = df['marca'].nunique()

print(marcas)
```

```python
# Criar ranking top 10 marcas mais valiosas

marcaTop10 = round(df.groupby('marca')['valor'].mean().sort_values(ascending=False).head(10))

print(marcaTop10)

```
