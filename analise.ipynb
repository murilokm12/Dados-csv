```python
import pandas as pd

# Criar um data frame(trazer os dados para a memória do python)

df = pd.read_csv('dados.csv')

display(df.head(2))


print('--------------------------------')
# print()
# print('def.info:')
print()

# info - apresenta um resumo do "dataset"
print(df.info)

print()
print('---------------------------------\n')

# 
print(df.describe())

# Mostra total de registros e colunas
print(df.shape)

print('\n')

print(df.columns)


```

```python
# Consultar apenas uma coluna 
marca =  df['marca']

print(marca)

print('\n')

# Consulta direto no print
print(df['modelo'])

print('\n')

# Consultar mais que uma coluna
marca_modelo_cor                       = df[['marca','modelo']]

display(marca_modelo_cor)
```

```python
# Criar filtros

# quero saber quantos carros não pagam mais ipva no meu estoque
noIpva = df[df['ano'] < 2006 ]

display(noIpva)

# Apresente os carros que custam mais de R$ 20.000,00
preco = df[df['valor'] > 20000]

display(preco)

# Apresente apenas os carro da chevrolet
carrosChlet = df[df['marca'] == 'chevrolet']

display(carrosChlet)

```
